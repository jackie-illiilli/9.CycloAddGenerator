```python
import numpy as np
import pandas as pd
```

```python
Mes_DeltaGa = [9.49758179, 8.62326782, 7.48601169, 7.62861448, 3.91878788, 5.074297631, 10.28538731, 9.619189363]
Ratio_4p = [0.78829816, 0.80559277, 0.78750361, 0.78012002, 0.74553303, 0.74800473, 0.822026983, 0.827569947]
from matplotlib import pyplot as plt
from sklearn.linear_model import LinearRegression
print(pd.DataFrame({'Ratio_4p': Ratio_4p, 'Mes_DeltaGa': Mes_DeltaGa}).corr(method="pearson"))

# Fit a linear regression model
model = LinearRegression()
X = np.array(Ratio_4p).reshape(-1, 1)
y = np.array(Mes_DeltaGa)
model.fit(X, y)
plt.figure(figsize=(5,5))
# Plot the data and the regression line
plt.scatter(X, y, color='blue', label='Data')
plt.plot(X, model.predict(X), color='red', label='Regression Line')
# plt.xlabel('Ratio_4p')
# plt.ylabel('Mes_DeltaGa')
# plt.legend()
# plt.show()

plt.scatter(Ratio_4p, Mes_DeltaGa)
plt.xticks(fontsize=20)
plt.yticks(fontsize=20)
plt.show()
```
